# Python `math` Module

## Introduction

The `math` module provides access to mathematical functions implemented in C for accuracy and performance. Use `math` when you need:

- Accurate floating-point operations (e.g., `fsum`)
- Special functions like `factorial`, `gcd`, `comb`, `perm`
- Trigonometric, logarithmic, and power functions

Import it using:

```python
import math
```

## Constants & Basic Functions

Important constants:
- `math.pi`, `math.e`, `math.tau`, `math.inf`, `math.nan`

Basic functions include `ceil`, `floor`, `trunc`, `fabs`, `factorial`, `isfinite`.

### Example: constants and basic functions

````python
import math

print('pi:', math.pi)
print('e:', math.e)
print('tau:', math.tau)
print('inf:', math.inf)
print('nan:', math.nan)

# Basic functions
print('ceil(4.1) ->', math.ceil(4.1))
print('floor(4.9) ->', math.floor(4.9))
print('trunc(4.9) ->', math.trunc(4.9))
print('fabs(-3.14) ->', math.fabs(-3.14))
print('factorial(5) ->', math.factorial(5))
print('isfinite(10) ->', math.isfinite(10))
print('isfinite(math.inf) ->', math.isfinite(math.inf))
````

## Powers & Roots

Functions:
- `math.pow(x, y)` — returns float x**y
- `math.sqrt(x)` — square root
- `math.isclose(a, b, rel_tol=1e-09, abs_tol=0.0)` — compare floats

### Example: powers and roots

````python
print('pow(2,3)=', math.pow(2,3))
print('2**3 =', 2**3)
print('sqrt(49)=', math.sqrt(49))

# isclose usage
print('isclose(0.1+0.2, 0.3) ->', math.isclose(0.1+0.2, 0.3, rel_tol=1e-9))
````

## Logarithms

Functions:
- `math.log(x)` — natural log (base e)
- `math.log10(x)` — log base 10
- `math.log2(x)` — log base 2
- `math.log1p(x)` — log(1+x) for better precision with small x

### Example: logarithms

````python
print('log(e)=', math.log(math.e))
print('log10(1000)=', math.log10(1000))
print('log2(8)=', math.log2(8))
print('log1p(1e-9)=', math.log1p(1e-9))
````

## Trigonometry

All trig functions use **radians**. Convert degrees⇄radians using `math.radians` and `math.degrees`.

Functions:
- `sin`, `cos`, `tan`
- `asin`, `acos`, `atan` (inverse)
- `sinh`, `cosh`, `tanh` (hyperbolic)

### Example: trigonometry

````python
# degrees to radians and back
print('radians(180)=', math.radians(180))
print('degrees(pi)=', math.degrees(math.pi))

# trig values
print('sin(pi/2)=', math.sin(math.pi/2))
print('cos(0)=', math.cos(0))
print('tan(pi/4)=', math.tan(math.pi/4))

# inverse trig
print('asin(1)=', math.degrees(math.asin(1)))

# hyperbolic
print('sinh(1)=', math.sinh(1))
````

##### Solutions

Below are solutions to the consolidated exercises. Read through, run the cells, and experiment by modifying values.

````python
# Solutions to selected exercises

import math

# 1. Area of circle
def area_circle(r):
    return math.pi * r * r

print('Area r=7:', area_circle(7))

# 2. Convert 120 deg -> radians and sin
print('sin(120°)=', math.sin(math.radians(120)))

# 3. GCD of 270 and 192
print('gcd(270,192)=', math.gcd(270,192))

# 4. 10!
print('10! =', math.factorial(10))

# 5. comb and perm
print('comb(12,4)=', math.comb(12,4))
print('perm(12,4)=', math.perm(12,4))

# 6. log1p
print('log1p(1e-9)=', math.log1p(1e-9))
print('log(1+1e-9)=', math.log(1+1e-9))

# 7. isclose example
print('isclose sum([0.1]*10, 1.0)=', math.isclose(sum([0.1]*10), 1.0, rel_tol=1e-9))
print('fsum([0.1]*10)=', math.fsum([0.1]*10))

# 8. amortization monthly payment
# Formula: M = P * r / (1 - (1+r)**-n)
# where r = monthly interest rate, n = months

def monthly_payment(P, annual_rate, years):
    r = annual_rate/12
    n = years*12
    return P * r / (1 - math.pow(1 + r, -n))

print('Monthly payment example:', monthly_payment(100000, 0.06, 15))

# 9. fsum precision
print('sum([0.1]*1000)=', sum([0.1]*1000))
print('fsum([0.1]*1000)=', math.fsum([0.1]*1000))

# 10. password strength metric (simple)
def password_strength_score(password):
    # score based on length and variety; use log to scale
    kinds = 0
    if any(c.islower() for c in password):
        kinds += 1
    if any(c.isupper() for c in password):
        kinds += 1
    if any(c.isdigit() for c in password):
        kinds += 1
    if any(not c.isalnum() for c in password):
        kinds += 1
    length = len(password)
    # simple metric
    return math.log1p(length) * kinds

print('password score for Abc123! =', password_strength_score('Abc123!'))

# 11. sqrt vs pow
print('sqrt(16)=', math.sqrt(16))
print('pow(2,4)=', math.pow(2,4))

# 12. batch convert angles
angles = [0, 30, 45, 60, 90]
print('cosines:', [math.cos(math.radians(a)) for a in angles])

# 13. prod
print('prod([1,2,3,4])=', math.prod([1,2,3,4]))

# 14. copysign
print('copysign(5, -2)=', math.copysign(5, -2))

# 15. isprime using isqrt

def is_prime(n):
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    r = math.isqrt(n)
    for i in range(3, r+1, 2):
        if n % i == 0:
            return False
    return True

print('is_prime(97)=', is_prime(97))
````
