## 1. Introduction

Strings are sequences of characters. In Python they are **immutable** (cannot change in place). Strings support indexing, slicing, iteration and many built-in methods.

```python
s = 'Hello, world!'
```

### 1.1 Creating Strings
- Single quotes: `'hello'`
- Double quotes: `"hello"`
- Triple quotes for multi-line strings:
```python
text = '''Line1
Line2'''
```

````python
# Examples of creation
s1 = 'hello'
s2 = "world"
s3 = '''This is
multiline''' 
print(s1, s2)
print(s3)
````

### 1.2 Immutability & Concatenation
- Strings are immutable: you cannot change a character in place.
- You can create new strings by concatenation or formatting.

```python
s = 'hello'
# s[0] = 'H'  # Error
s2 = 'H' + s[1:]
```

````python
s = 'hello'
try:
    s[0] = 'H'
except TypeError as e:
    print('Error:', e)

s2 = 'H' + s[1:]
print(s2)
````

## 2. Indexing & Slicing

- Indexing starts at 0. Use negative indices from the end.
- Slicing: `s[start:stop:step]` (stop is exclusive)

Examples and tips included below.

````python
s = 'Python'
print('s[0]=', s[0])
print('s[-1]=', s[-1])
print('s[1:4]=', s[1:4])
print('s[:3]=', s[:3])
print('s[::2]=', s[::2])
print('Reverse:', s[::-1])
````

## 3. Case Conversion Methods
- `upper()`, `lower()`, `title()`, `capitalize()`, `swapcase()`, `casefold()`

`casefold()` is recommended for aggressive case-insensitive matching (useful for Unicode).

````python
s = 'PyThOn Programming'
print('upper ->', s.upper())
print('lower ->', s.lower())
print('title ->', s.title())     #Capitalizes each word.
print('capitalize ->', s.capitalize())   #Capitalizes first letter of the string.
print('swapcase ->', s.swapcase())     #Swaps the case of each letter.
print('casefold ->', s.casefold())      #More aggressive lowercasing for caseless matching.

s = "Straße"
print(s.casefold())   # strasse
````

## 4. Searching & Finding
- `find()` returns -1 if not found, `index()` raises `ValueError`.
- `rfind()` and `rindex()` search from the right.
- `count()` returns occurrences.

**Note:** prefer `in` for boolean checks: `'sub' in s`.

````python
s = 'banana'
print('find a ->', s.find('a'))   #Returns index of substring (returns -1 if not found)
````

````python
s = "Python Programming"
print(s.find("Pro"))     # 7
print(s.find("xyz"))     # -1
````

````python
print('index a ->', s.index('a'))   #Raises ValueError if substring not found
print('rfind a ->', s.rfind('a'))  #Finds substring from end.
print('count a ->', s.count('a'))
print("'nan' in s ->", 'nan' in s)   #Membership test
````

## 5. Splitting & Joining
- `split(sep)` turns string into list
- `join(iterable)` merges list into string
- `partition(sep)` returns tuple (before, sep, after)

Teaching tip: show how split without args splits on whitespace and handles multiple spaces.

````python
text = '  Python   is   fun  '
print('split ->', text.split())   #Splits on whitespace by default
print('split(" ") ->', text.split(' '))   #Splits on space character
words = text.split()
print('join ->', '-'.join(words))   #Joins words with '-'
print('partition ->', 'hello-world'.partition('-'))  #Splits into 3 parts: before, sep, after
````

## 6. Strip & Padding
- `strip()`, `lstrip()`, `rstrip()`
- `zfill(width)` pads numeric strings

Exercise: clean CSV fields and pad IDs.

````python
s = '   hello   '
print('strip:', s.strip())    #Removes leading and trailing whitespace
print('lstrip:', s.lstrip())  #Removes leading whitespace
print('rstrip:', s.rstrip())   #Removes trailing whitespace
print('zfill:', '42'.zfill(5))   #Pads string on the left with zeros to a total width
````

````python
"---hello---".lstrip("-")    # 'hello---'
"---hello---".rstrip("-")    # '---hello'
"---hello---".strip("-")     # 'hello'
````

## 8. Formatting Strings
- `%` formatting (old)
- `str.format()` (modern)
- f-strings (recommended in Python 3.6+)

Examples below show when to use each and teaching notes.

````python
name = 'Shahma'
age = 30
# percent
print('Name: %s, Age: %d' % (name, age)) 
# format
print('Name: {}, Age: {}'.format(name, age))
# f-string
print(f'Name: {name}, Age: {age}')

# formatting options
pi = 3.1415926535
print(f'Pi to 3 decimals: {pi:.3f}')
````

````python
print("My name is {}".format("Shahma"))
````

````python
name = "Shahma"
age = 25
print(f"My name is {name} and I am {age} years old.")  #F-strings
````

## 9. Validation Methods (Boolean checks)
- `isalpha()`, `isdigit()`, `isalnum()`, `isspace()`, `islower()`, `isupper()`, `istitle()`

Use-cases: validating user input like IDs, names, phone numbers (with caution!).

````python
print('abc'.isalpha(),   #Checks if all characters are alphabetic
       'abc123'.isalnum(),  #Checks if all characters are alphanumeric
         '123'.isdigit(),    #Checks if all characters are digits
         ' \t\n'.isspace(),  #Checks if all characters are whitespace
           '   '.isspace())  #Checks if all characters are whitespace
print('Title'.istitle(), 'HELLO'.isupper(), 'hello'.islower())

````

## 11. Encoding & Bytes
- `encode()` turns str -> bytes, `decode()` reverses
- Useful when reading/writing files, networking

Example:

````python
s = 'café'
b = s.encode('utf-8')    # Encode to bytes using UTF-8
print('bytes:', b)
print('decoded:', b.decode('utf-8'))
````

````python
"hello".encode() 
# b'hello'
````

## 12. Iteration & Performance Tips
- Iterating over characters: `for ch in s:`
- Avoid repeated concatenation in loops; use `join()` for building strings efficiently.

Teaching tip: demonstrate O(n^2) behavior of repeated `+=` concatenation vs `append` + `join`.

````python
# Bad: repeated concatenation
s = ''
for i in range(1000):
    s += 'x'

# Good: list + join
parts = []
for i in range(1000):
    parts.append('x')
s2 = ''.join(parts)
print(len(s), len(s2))
````

## 13. Practical Examples & Mini-Projects
1. Email domain extractor
2. Palindrome checker
3. Word frequency counter
4. Simple template renderer

Examples below:

````python
# 1. Email domain extractor
def extract_domain(email):
    try:
        return email.split('@')[1].lower()
    except IndexError:
        return None

print(extract_domain('user@example.com'))
print(extract_domain('invalid-email'))

# 2. Palindrome checker
def is_palindrome(s):
    s2 = ''.join(ch.lower() for ch in s if ch.isalnum())
    return s2 == s2[::-1]

print(is_palindrome('A man, a plan, a canal: Panama'))

# 3. Word frequency
from collections import Counter

def word_freq(text):
    words = [w.lower().strip('.,!?') for w in text.split()]
    return Counter(words)

print(word_freq('This is a test. This test is simple.'))
````

## 15. Solutions (Selected)

Below are solutions for a subset of the exercises. Encourage students to attempt before opening solutions.

````python
# Solution 1: Count vowels
def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in 'aeiou')

print(count_vowels('Education'))

# Solution 2: Validate password
def validate_password(pw):
    if len(pw) < 8:
        return False
    has_digit = any(ch.isdigit() for ch in pw)
    has_alpha = any(ch.isalpha() for ch in pw)
    return has_digit and has_alpha

print(validate_password('abc123'))
print(validate_password('StrongPass1'))

# Solution 3: Reverse each word

def reverse_each_word(sentence):
    return ' '.join(word[::-1] for word in sentence.split())

print(reverse_each_word('python is fun'))

# Solution 4: snake_case -> CamelCase

def snake_to_camel(s):
    parts = s.split('_')
    return ''.join(p.capitalize() for p in parts if p)

print(snake_to_camel('this_is_snake'))

# Solution 5: Extract numbers
import re

def extract_numbers(s):
    return [int(x) for x in re.findall(r'\d+', s)]

print(extract_numbers('room 123, floor 2'))
````
