# **PANDAS LIBRARY IN PYTHON**

* Pandas is a powerful open-source Python library used for data manipulation, analysis, and cleaning.

* It provides easy-to-use data structures like:

      * Series → 1-dimensional labeled data

      * DataFrame → 2-dimensional table (rows & columns)

* Pandas is built on top of NumPy and is widely used in data science, ML, analytics, and data preprocessing.
* Pandas allows us to perform operations such as:



```
Load Data
   ↓
Inspect Data
   ↓
Clean Data
   ↓
Transform Data
   ↓
Analyze Data
   ↓
Prepare Data for ML
```



### Why Pandas?

* Easy to read/write datasets (CSV, Excel, SQL, JSON)

* Fast data manipulation

* Handles missing data

* Label-based indexing

* Built-in statistical operations

* Integration with NumPy, Matplotlib, Scikit-Learn

#### Filtering Data

````python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Rahul", "Anu", "John"],
    "Age": [20, 21, 20, 22, 21],
    "Course": [
        "Data Science",
        "Python",
        "Data Science",
        "Python",
        "Data Science"
    ],
    "Marks": [85, 78, 92, 88, 76],
    "Attendance": [90, 85, 95, 88, 80]
}

df = pd.DataFrame(data)

print(df)
````

````python
# students with marks greater than 80.

high_marks = df[df["Marks"] > 80]

print(high_marks)
````

````python
# Students with marks greater than 80 and attendance greater than 90:
result = df[
    (df["Marks"] > 80) &
    (df["Attendance"] > 90)
]

print(result)
````

````python
result = df[
    (df["Marks"] > 90) |
    (df["Attendance"] > 90)
]

print(result)
````

Adding a New Column

````python
df["Result"] = "Pass"
print(df)
````

````python
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
df
````

Updating a Column

````python
df["Marks"] = df["Marks"] + 5
````

Removing a Column

### Data Cleaning & Manipulation with Pandas

* Data cleaning involves preparing raw data by correcting or removing incorrect, incomplete, or irrelevant data.

#### 1. Handling Missing Values

````python
data = {
    'Name': ['Achu', 'John', 'Meera', 'Ravi'],
    'Age': [22, None, 23, None],
    'Marks': [85, 90, None, 70]
}

df = pd.DataFrame(data)
print(df)
````

````python
df.info()
````

Check Missing Values

````python
df.isnull()
````

````python
df.isnull().sum()
````

````python
df.isnull().sum().sum()
````

````python
df.isna()          # True for missing values
df.isna().sum()    # Count missing values column-wise
````

````python
print(df.isnull())
````

dropna() – Remove missing values

````python
df_drop = df.dropna()     # Remove rows with any NaN:
df_drop
````

````python
df_drop_col = df.dropna(axis=1)  # Remove columns with NaN:
````

````python
df_drop_col
````

fillna() – Replace missing values

````python
df_filled = df.fillna(0)       # Fill with specific value:
````

````python
df_filled
````

````python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
````

````python
df
````

````python
print(df.duplicated())
````

````python
print(df.duplicated())
````

````python
df = df.rename(
    columns={"Marks": "Score"}
)
df
````

````python
df["Age"] = df["Age"].astype(int)
````

````python
df["Age"] = df["Age"].astype(int)
````

## GroupBy

````python
df
````

````python
# the average marks for each course.
result = df.groupby("Age")["Score"].mean()
print(result)
````

````python
result = df.groupby("Age")["Score"].agg(
    ["mean", "max", "min", "count"]
)
print(result)
````

````python
def grade(marks):

    if marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    else:
        return "D"


df["Grade"] = df["Score"].apply(grade)

print(df)
````

## Combining DataFrames — concat()

````python
df1 = pd.DataFrame({
    "Name": ["Alice", "Bob"],
    "Marks": [85, 78]
})

df2 = pd.DataFrame({
    "Name": ["Rahul", "Anu"],
    "Marks": [92, 88]
})

result = pd.concat([df1, df2])
print(result)
````

````python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Rahul"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [85, 78, 92]
})

result = pd.merge(
    students,
    marks,
    on="ID"
)

print(result)
````
