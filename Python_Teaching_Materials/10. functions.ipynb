### Positional Arguments

In positional arguments, values are assigned according to their position.

* These are the most common type.

* The order of arguments matters — the first value goes to the first parameter, the second to the second, and so on.



```
def function_name(parameter1, parameter2):
    statements

function_name(value1, value2)
```



````python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Aswathi", 27)                       # Here "Aswathi" → name, and 27 → age.
student(27,"Aswathi")                        # Here name parameter will get 27 and age parameter will get "Aswathi" as value
````

### Keyword Arguments

In keyword arguments, parameter names are specified during function call.



```

```



* You can specify which value goes to which parameter by name, regardless of order.

````python
def student(place,name, age):
    print("Name:", name)
    print("Age:", age)
    print("Place:", place)
    print("\n")

student("ksd",age=22, name="Aswathi")
# Output is the same — order doesn’t matter.
student("ksd",name="Ammu",age=25)
````

### Variable Length Arguments (*args)

Allows passing multiple positional arguments.



```
def function(*args):
```



* If you don’t know how many arguments will be passed, use *args.

* It collects extra positional arguments into a tuple.

````python
def add(*numbers):
    total = 0
    for n in numbers:
        total += n
    print("Sum =", total)

add(2, 3)
add(2, 3, 4, 5)
````

````python
def show_items(*items):
    print("Items are:")
    for i in items:
        print("*", i)

show_items("Apple", "Banana", "Cherry",89,67)
````

Combine Fixed and Variable Arguments

````python
def greet(msg, *names):
    for name in names:
        print(msg, name)

greet("Hello", "Aswathi", "Anu", "Rahul")       # msg parameter gets - "Hello" and all other values will be assigned to *names
````

## Recursive Functions

Function calling itself.

````python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
````

````python
def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n - 1)
countdown(5)
````

## Lambda Functions

Anonymous functions.



```
lambda arguments : expression
```



* Lambda functions in Python are anonymous, single-expression functions defined using the lambda keyword

* They're compact and often used for short, throwaway operations.



````python
square = lambda x: x * x
print(square(5))
````

````python
add = lambda a, b: a + b
print(add(10, 20))
````

- No function name
- Returns the result of the expression automatically


````python
add = lambda x, y: x + y
print(add(3, 5))
````

Common Use Cases:

- With map():    Apply a function to every item in a list


````python
nums = [1, 2, 3]
squares = list(map(lambda x: x**2, nums))
print(squares)
````

- With filter(): Filter items based on a condition

````python
nums = [1, 2, 3, 4]
evens = list(filter(lambda x: x % 2 == 0, nums))
evens
````

## Scope of Variables

`LEGB Rule`

Python searches variables in order:

- Local
- Enclosing
- Global
- Built-in

>  What is Scope?

Scope defines **where a variable can be accessed**.

Python follows a strict order when looking for variables.


````python
x = 100

def test():
    x = 50
    print(x)
test()

print(x)
````

````python
# ## Example 1: Local Scope
def func():
    x = "local"
    print("Inside function:", x)

func()
````

````python
## Example 2: Enclosing Scope
def outer():
    x = "enclosing"

    def inner():
        print("Inside inner:", x)

    inner()

outer()
````

````python
## Example 3: Global Scope
x = "global"

def func():
    print("Inside function:", x)

func()
print("Outside function:", x)
````

````python
## Example 4: Full LEGB Rule
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print("Inner:", x)

    inner()
    print("Outer:", x)

outer()
print("Global:", x)
````

````python

x = "global"

def outer():
    x = "enclosing"

    def inner():
        # Try commenting/uncommenting below line
        # x = "local"
        print("Inner:", x)

    inner()

outer()
````


> Special Keywords

- `global` → modify global variable
- `nonlocal` → modify enclosing variable


````python

x = "global"

def outer():
    x = "enclosing"

    def inner():
        nonlocal x
        x = "modified enclosing"

    inner()
    print("Outer:", x)

outer()
````

## Practical Examples & Mini-Projects
1. Email domain extractor
2. Palindrome checker
3. Word frequency counter
4. Simple template renderer

Examples below:

````python
# 1. Email domain extractor
def extract_domain(email):
    try:
        return email.split('@')[1].lower()
    except IndexError:
        return None

print(extract_domain('user@example.com'))
print(extract_domain('invalid-email'))

# 2. Palindrome checker
def is_palindrome(s):
    s2 = ''.join(ch.lower() for ch in s if ch.isalnum())
    return s2 == s2[::-1]

print(is_palindrome('A man, a plan, a canal: Panama'))

# 3. Word frequency
from collections import Counter

def word_freq(text):
    words = [w.lower().strip('.,!?') for w in text.split()]
    return Counter(words)

print(word_freq('This is a test. This test is simple.'))
````

````python
# Solution 1: Count vowels
def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in 'aeiou')

print(count_vowels('Education'))

# Solution 2: Validate password
def validate_password(pw):
    if len(pw) < 8:
        return False
    has_digit = any(ch.isdigit() for ch in pw)
    has_alpha = any(ch.isalpha() for ch in pw)
    return has_digit and has_alpha

print(validate_password('abc123'))
print(validate_password('StrongPass1'))

# Solution 3: Reverse each word

def reverse_each_word(sentence):
    return ' '.join(word[::-1] for word in sentence.split())

print(reverse_each_word('python is fun'))

# Solution 4: snake_case -> CamelCase

def snake_to_camel(s):
    parts = s.split('_')
    return ''.join(p.capitalize() for p in parts if p)

print(snake_to_camel('this_is_snake'))

# Solution 5: Extract numbers
import re

def extract_numbers(s):
    return [int(x) for x in re.findall(r'\d+', s)]

print(extract_numbers('room 123, floor 2'))
````
