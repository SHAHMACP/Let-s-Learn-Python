### Indexing & Slicing

- Indexing starts at 0. Use negative indices from the end.
- Slicing: `s[start:stop:step]` (stop is exclusive)

Examples and tips included below.

````python
s = 'Python'
print('s[0]=', s[0])     #first char
print('s[-1]=', s[-1])     #last char
print('s[1:4]=', s[1:4])   #from char from 1st index to 4 excluded
print('s[:3]=', s[:3])   #upto 3rd, excluding
print('s[::2]=', s[::2])   #alternating char
print('Reverse:', s[::-1])  #reversing
````

````python
name="Python Programming"

print(name[0:9])                          # STARTING POSITION 0 AND STOP POSITION 9(EXCLUSIVE)
print(name[:])                             # IF STARTING AND STOP INDEX IS EMPTY THAT MEANS RETURN THE ENTIRE STRING

print(name[9])                            # RETURNS THE 9 TH CHARACTER
print(name[2])

print(name[3:])                            # STARTING FROM 3 POSITION TILL END
print(name[:7])                            # STARTING IS EMPTY SO IT WILL START FROM BEGINNING AND STOP AT 7 POSITION(7 position value will not be included)

print(name[0:-1])                          # START AT 0 TH POSITION AND STOP AT LAST INDEX(EXCLUSIVE)
print(name[-4:-1])

print(name[0:9:2])                         # START 0 STOP 9 STEP 2-IT WILL SKIP ONE ELEMENT
print(name[3:8:3])

print(name[::-1])
print(name[-3:-7:-1])
print(name[-2:-10:-2])
````

### Case Conversion Methods
- `upper()`, `lower()`, `title()`, `capitalize()`, `swapcase()`, `casefold()`

`casefold()` is recommended for aggressive case-insensitive matching (useful for Unicode).

````python
s = 'PyThOn Programming'
print('upper ->', s.upper())
print('lower ->', s.lower())
print('title ->', s.title())     #Capitalizes each word.
print('capitalize ->', s.capitalize())   #Capitalizes first letter of the string.
print('swapcase ->', s.swapcase())     #Swaps the case of each letter.
print('casefold ->', s.casefold())      #More aggressive lowercasing for caseless matching.

s = "Straße"
print(s.casefold())   # strasse
````

````python
text="HEllo world"
print(text)

print(text.upper())        # CONVERTS ALL CHARACTERS TO UPPERCASE

print(text.lower())        # CONVERTS ALL CHARACTERS TO LOWERCASE

print(text.capitalize())    # Capitalizes the first character

print(text.title())         # Converts first letter of each word to uppercase

print(text.swapcase())      # Swaps case of all letters
````

### Searching & Finding
- `find()` returns -1 if not found, `index()` raises `ValueError`.
- `rfind()` and `rindex()` search from the right.
- `count()` returns occurrences.

**Note:** prefer `in` for boolean checks: `'sub' in s`.

````python
s = 'banana'
s.find('a')   #Returns index of substring (returns -1 if not found)
````

````python
s = "Python Programming"
print(s.find("Pro"))     # 7
print(s.find("xyz"))     # -1
````

````python
print('index a ->', s.index('a'))   #Raises ValueError if substring not found
print('rfind a ->', s.rfind('a'))  #Finds substring from end.
print('count a ->', s.count('a'))
print("'nan' in s ->", 'nan' in s)   #Membership test
````

````python
text="hello welcome to python programming python"

print(text.startswith("hello"))                      # Checks if string starts with substring hello, return true or false
print(text.endswith("hello"))                       #   Checks if string ends with substring hello, return true or false

print(text.find("o"))                          # Returns index of first occurrence of substring (-1 if not found)
print(text.find("Z"))
print(text.find("python"))

print(text.index('o'))                             # Returns index of first occurrence (raises error if not found)

print(text.count('o'))                            # Counts occurrences of substring 'o'

print(text.isalpha())                            # Checks if all characters are letters , if yes returns true otherwise false
print(text.isdigit())                             # Checks if all characters are digits, if yes returns true otherwise false
print(text.isspace())                             # Checks if all characters are whitespace, if yes returns true otherwise false

print(text.islower())                              # Checks if all characters are lowercase
print(text.isupper())                              # Checks if all characters are uppercase
print(text.isalnum())                               # Checks if all characters are alphanumeric
````

### Splitting & Joining
- `split(sep)` turns string into list
- `join(iterable)` merges list into string
- `partition(sep)` returns tuple (before, sep, after)

Teaching tip: show how split without args splits on whitespace and handles multiple spaces.

````python
text = '  Python   is   fun  '
print('split ->', text.split())   #Splits on whitespace by default
print('split(" ") ->', text.split(' '))   #Splits on space character
words = text.split()
print('join ->', '-'.join(words))   #Joins words with '-'
print('partition ->', 'hello-world'.partition('-'))  #Splits into 3 parts: before, sep, after
````

### Iteration & Performance Tips
- Iterating over characters: `for ch in s:`
- Avoid repeated concatenation in loops; use `join()` for building strings efficiently.

Teaching tip: demonstrate O(n^2) behavior of repeated `+=` concatenation vs `append` + `join`.

````python
# Bad: repeated concatenation
s = ''
for i in range(1000):
    s += 'x'
print(s)
print(len(s))
````

````python
# Good: list + join
parts = []
for i in range(1000):
    parts.append('x')
s2 = ''.join(parts)
print(s2)
print(len(s2))
````

### Practical Examples & Mini-Projects
1. Email domain extractor
2. Palindrome checker
3. Word frequency counter
4. Simple template renderer

Examples below:

````python
# 1. Email domain extractor
def extract_domain(email):
    try:
        return email.split('@')[1].lower()
    except IndexError:
        return None

print(extract_domain('user@example.com'))
print(extract_domain('invalid-email'))

# 2. Palindrome checker
def is_palindrome(s):
    s2 = ''.join(ch.lower() for ch in s if ch.isalnum())
    return s2 == s2[::-1]

print(is_palindrome('A man, a plan, a canal: Panama'))

# 3. Word frequency
from collections import Counter

def word_freq(text):
    words = [w.lower().strip('.,!?') for w in text.split()]
    return Counter(words)

print(word_freq('This is a test. This test is simple.'))
````

4. Convert snake_case to CamelCase.
5. Extract all numbers from a string and return as list of ints.
6. Check if two strings are anagrams.
7. Remove duplicate words from a sentence (preserve order).

#### Solutions (Selected)

Below are solutions for a subset of the exercises. Encourage students to attempt before opening solutions.

````python
# Solution 1: Count vowels
def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in 'aeiou')

print(count_vowels('Education'))

# Solution 2: Validate password
def validate_password(pw):
    if len(pw) < 8:
        return False
    has_digit = any(ch.isdigit() for ch in pw)
    has_alpha = any(ch.isalpha() for ch in pw)
    return has_digit and has_alpha

print(validate_password('abc123'))
print(validate_password('StrongPass1'))

# Solution 3: Reverse each word

def reverse_each_word(sentence):
    return ' '.join(word[::-1] for word in sentence.split())

print(reverse_each_word('python is fun'))

# Solution 4: snake_case -> CamelCase

def snake_to_camel(s):
    parts = s.split('_')
    return ''.join(p.capitalize() for p in parts if p)

print(snake_to_camel('this_is_snake'))

# Solution 5: Extract numbers
import re

def extract_numbers(s):
    return [int(x) for x in re.findall(r'\d+', s)]

print(extract_numbers('room 123, floor 2'))
````

### STRING TO INTEGER & INTEGER TO STRING

````python
a="abc123"
x="123"                # STRING
print(type(x))


y=int(x)              # CONVERTED THE STRING TO INTEGER  (ONLY POSSIBLE IF THE STRING IS CONTAINS AN INTEGER VALUE)
print(y)
print(type(y))         #INTEGER


#z=int(a)            # NOT POSSIBLE BECAUSE THE STRING CONTAINS LETTERS ALSO

num=1234
string=str(num)        # converted the integer to string
print(string)
print(type(string))
````

````python
# Sometimes you need to convert one type to another (casting).

x = "42"
y = int(x)           # String to int
print("Converted:", y, type(y))

z = 77
w = float(z)         # Int to float
print(w, type(w))

n = 3.14159
str_n = str(n)       # Float to string
print(str_n, type(str_n))

# Boolean conversions
flag = 1
print(bool(flag))    # True, because 1 is truthy
flag = 0
print(bool(flag))    # False, because 0 is falsy

# Points to remember:
# - Use int(), float(), str(), bool() for conversions.
# - Invalid conversions (e.g., int("hello")) give errors.
````
